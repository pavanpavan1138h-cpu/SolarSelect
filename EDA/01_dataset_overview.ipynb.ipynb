```python
from pathlib import Path

data_path = Path("../data/raw/archive")
print(data_path.resolve())
print(list(data_path.glob("*.csv")))
```

```python
import pandas as pd

files = {
    "plant1_generation": "Plant_1_Generation_Data.csv",
    "plant1_weather": "Plant_1_Weather_Sensor_Data.csv",
    "plant2_generation": "Plant_2_Generation_Data.csv",
    "plant2_weather": "Plant_2_Weather_Sensor_Data.csv"
}

datasets = {
    name: pd.read_csv(data_path / filename)
    for name, filename in files.items()
}

print("All datasets loaded successfully!")
```

```python
for name, filename in files.items():
    raw_df = pd.read_csv(data_path / filename)

    print(f"\n{'=' * 40}")
    print(name)
    print("First 5 original timestamps:")
    print(raw_df["DATE_TIME"].head().to_list())
    print("Last 5 original timestamps:")
    print(raw_df["DATE_TIME"].tail().to_list())
```

```python
date_formats = {
    "plant1_generation": "%d-%m-%Y %H:%M",
    "plant1_weather": "%Y-%m-%d %H:%M:%S",
    "plant2_generation": "%Y-%m-%d %H:%M:%S",
    "plant2_weather": "%Y-%m-%d %H:%M:%S"
}

for name, filename in files.items():
    df = pd.read_csv(data_path / filename)

    df["DATE_TIME"] = pd.to_datetime(
        df["DATE_TIME"],
        format=date_formats[name]
    )

    datasets[name] = df

    print(
        name,
        df["DATE_TIME"].min(),
        "to",
        df["DATE_TIME"].max()
    )
```

```python
for plant in ["plant1", "plant2"]:
    generation = datasets[f"{plant}_generation"]
    weather = datasets[f"{plant}_weather"]

    gen_times = set(generation["DATE_TIME"].unique())
    weather_times = set(weather["DATE_TIME"].unique())

    print(f"\n{plant}")
    print("Unique generation timestamps:", len(gen_times))
    print("Unique weather timestamps:", len(weather_times))
    print("Matching timestamps:", len(gen_times & weather_times))
    print("Generation timestamps without weather:", len(gen_times - weather_times))
```

```python
plant1_gen_times = set(
    datasets["plant1_generation"]["DATE_TIME"].unique()
)

plant1_weather_times = set(
    datasets["plant1_weather"]["DATE_TIME"].unique()
)

missing_times = plant1_gen_times - plant1_weather_times

print("Unmatched generation timestamps:")
print(missing_times)
```

```python
for plant in ["plant1", "plant2"]:
    df = datasets[f"{plant}_generation"]

    print(f"\n{plant}")
    print(df.groupby("DATE_TIME").size().value_counts().sort_index())
```

```python
for plant in ["plant1", "plant2"]:
    df = datasets[f"{plant}_generation"]

    print(f"\n{plant}")
    print("Unique inverters:", df["SOURCE_KEY"].nunique())
    print("Records per inverter:")
    display(df["SOURCE_KEY"].value_counts())
```

```python
for plant in ["plant1", "plant2"]:
    df = datasets[f"{plant}_generation"]

    print(f"\n{'=' * 40}")
    print(plant)

    coverage = df.groupby("SOURCE_KEY")["DATE_TIME"].agg(
        ["min", "max", "count"]
    )

    display(coverage.sort_values("count"))
```

```python
for plant in ["plant1", "plant2"]:
    df = datasets[f"{plant}_generation"]

    print(f"\n{plant}")
    print("AC_POWER:")
    print("  Min:", df["AC_POWER"].min())
    print("  Max:", df["AC_POWER"].max())
    print("  Zero values:", (df["AC_POWER"] == 0).sum())
    print("  Negative values:", (df["AC_POWER"] < 0).sum())
```

```python
plant1_gen = datasets["plant1_generation"]
plant1_weather = datasets["plant1_weather"]

plant1_merged = plant1_gen.merge(
    plant1_weather[
        ["DATE_TIME", "AMBIENT_TEMPERATURE",
         "MODULE_TEMPERATURE", "IRRADIATION"]
    ],
    on="DATE_TIME",
    how="inner"
)

print("Generation rows:", len(plant1_gen))
print("Merged rows:", len(plant1_merged))
print("\nColumns:")
print(plant1_merged.columns.tolist())
```

```python
print(
    plant1_merged[
        ["IRRADIATION", "AC_POWER"]
    ].corr()
)
```

```python
display(
    plant1_merged[
        ["IRRADIATION", "AC_POWER"]
    ].describe()
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(
    plant1_merged["IRRADIATION"],
    plant1_merged["AC_POWER"],
    alpha=0.2
)

plt.xlabel("Irradiation")
plt.ylabel("AC Power")
plt.title("Irradiation vs AC Power — Plant 1")

plt.show()
```

```python
print(
    plant1_merged[
        [
            "AMBIENT_TEMPERATURE",
            "MODULE_TEMPERATURE",
            "IRRADIATION",
            "AC_POWER"
        ]
    ].corr()["AC_POWER"].sort_values(ascending=False)
)
```

```python
daytime = plant1_merged[plant1_merged["IRRADIATION"] > 0]

print("Total observations:", len(plant1_merged))
print("Daylight observations:", len(daytime))
print("Zero-irradiation observations:", (plant1_merged["IRRADIATION"] == 0).sum())
```
