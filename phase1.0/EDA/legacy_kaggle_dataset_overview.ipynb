```python
def iqr_outlier_summary(df, columns):
    results = []

    for column in columns:
        values = df[column].dropna()

        q1 = values.quantile(0.25)
        q3 = values.quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outliers = (
            (values < lower_bound) |
            (values > upper_bound)
        )

        results.append({
            "Column": column,
            "Q1": q1,
            "Q3": q3,
            "IQR": iqr,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Potential Outliers": int(outliers.sum()),
            "Outlier Percentage": round(
                outliers.mean() * 100, 2
            )
        })

    return pd.DataFrame(results)


print("PLANT 1 — GENERATION")
display(
    iqr_outlier_summary(
        datasets["plant1_generation"],
        ["AC_POWER", "DC_POWER", "DAILY_YIELD"]
    )
)

print("PLANT 2 — GENERATION")
display(
    iqr_outlier_summary(
        datasets["plant2_generation"],
        ["AC_POWER", "DC_POWER", "DAILY_YIELD"]
    )
)

print("PLANT 1 — WEATHER")
display(
    iqr_outlier_summary(
        datasets["plant1_weather"],
        weather_columns
    )
)

print("PLANT 2 — WEATHER")
display(
    iqr_outlier_summary(
        datasets["plant2_weather"],
        weather_columns
    )
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plant 1: generation + weather, matched by timestamp
p1 = datasets["plant1_generation"].merge(
    datasets["plant1_weather"],
    on=["DATE_TIME", "PLANT_ID"],
    how="inner",
    suffixes=("_gen", "_weather")
)

# Plant 2: generation + weather, matched by timestamp
p2 = datasets["plant2_generation"].merge(
    datasets["plant2_weather"],
    on=["DATE_TIME", "PLANT_ID"],
    how="inner",
    suffixes=("_gen", "_weather")
)

columns = [
    "AC_POWER",
    "DC_POWER",
    "DAILY_YIELD",
    "AMBIENT_TEMPERATURE",
    "MODULE_TEMPERATURE",
    "IRRADIATION"
]

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

for ax, df, plant_name in zip(
    axes,
    [p1, p2],
    ["Plant 1", "Plant 2"]
):
    correlation = df[columns].corr()

    sns.heatmap(
        correlation,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        center=0,
        ax=ax
    )

    ax.set_title(f"{plant_name} — Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, df, plant_name in zip(
    axes,
    [p1, p2],
    ["Plant 1", "Plant 2"]
):
    data = df.copy()
    data["DATE_TIME"] = pd.to_datetime(data["DATE_TIME"])
    data["HOUR"] = data["DATE_TIME"].dt.hour

    hourly_power = data.groupby("HOUR")["AC_POWER"].mean()

    ax.plot(
        hourly_power.index,
        hourly_power.values,
        marker="o"
    )

    ax.set_title(f"{plant_name} — Average Hourly Power")
    ax.set_xlabel("Hour of Day")
    ax.set_ylabel("Mean AC_POWER")
    ax.set_xticks(range(0, 24, 2))
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

for plant_name, df in [("Plant 1", p1), ("Plant 2", p2)]:
    data = df.copy()
    data["DATE_TIME"] = pd.to_datetime(data["DATE_TIME"])

    # Use irradiation to define daylight
    daylight = data["IRRADIATION"] > 0

    print(f"\n{plant_name}")
    print("Daylight rows:", int(daylight.sum()))
    print("Zero-irradiation rows:", int((~daylight).sum()))

    print("\nMean AC_POWER by irradiation condition:")
    print(
        data.groupby(daylight)["AC_POWER"]
        .agg(["count", "mean", "median"])
        .rename(index={False: "Zero irradiation", True: "Positive irradiation"})
    )

    print("\nMean AC_POWER by coverage status:")
    print(
        data.groupby("complete_coverage")["AC_POWER"]
        .agg(["count", "mean", "median"])
    )
```
